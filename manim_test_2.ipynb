````python
%%manim -qm CircleToSquare

class CircleToSquare(Scene):
    def construct(self):
        blue_circle = Circle(color=BLUE, fill_opacity=0.5)
        green_square = Square(color=GREEN, fill_opacity=0.8)
        self.play(Create(blue_circle))
        self.wait()
        
        self.play(Transform(blue_circle, green_square))
        self.wait()
````

````python
%%manim -qm HadamardGateAnimation
class HadamardGateAnimation(ThreeDScene):
    def construct(self):
        self.set_camera_orientation(phi=70 * DEGREES, theta=30 * DEGREES)
        axes = ThreeDAxes(x_range=[-1.5, 1.5], y_range=[-1.5, 1.5], z_range=[-1.5, 1.5])
        sphere = Sphere(radius=1, resolution=(24, 24), fill_opacity=0.15, checkerboard_colors=False)
        sphere.set_color(BLUE_E)
        
        label_0 = MathTex("|0\\rangle").next_to(axes.z_axis.get_end(), UP)
        label_1 = MathTex("|1\\rangle").next_to(axes.z_axis.get_start(), DOWN)
        self.add_fixed_orientation_mobjects(label_0, label_1)
        
        # Initial state |0> points along +z
        start_vec = np.array([0, 0, 1])
        # After Hadamard: |0> -> (|0>+|1>)/sqrt(2), Bloch vector points along +x
        end_vec = np.array([1, 0, 0])
        
        arrow = Arrow3D(start=ORIGIN, end=start_vec, color=YELLOW)
        
        self.play(Create(axes), Create(sphere))
        self.play(GrowFromCenter(arrow))
        self.wait(0.5)

        title = Text("Applying Hadamard Gate", font_size=28).to_corner(UL)
        self.add_fixed_in_frame_mobjects(title)
        self.play(Write(title))

        new_arrow = Arrow3D(start=ORIGIN, end=end_vec, color=GREEN)
        self.play(Transform(arrow, new_arrow), run_time=2)
        self.wait(1)

        state_tex = MathTex(
            "H|0\\rangle = \\frac{1}{\\sqrt{2}}(|0\\rangle + |1\\rangle)",
            font_size=32
        ).to_corner(DL)
        self.add_fixed_in_frame_mobjects(state_tex)
        self.play(Write(state_tex))
        self.wait(2)
````

````python
%%manim -qm TunnelingProblems

from manim import *

class TunnelingProblems(Scene):
    def construct(self):
        title = Text("Quantum Tunneling Through a Potential Barrier", font_size=32).to_edge(UP)
        self.play(Write(title))

        axes = Axes(x_range=[-6, 6, 1], y_range=[0, 3, 1], x_length=10, y_length=4,
                     axis_config={"include_tip": False}).shift(DOWN * 0.5)
        self.play(Create(axes))

        def barrier(x):
            return 2 if -1 <= x <= 1 else 0

        barrier_graph = axes.plot(barrier, x_range=[-6, 6], use_smoothing=False, color=RED)
        barrier_label = Text("V(x)", font_size=24, color=RED).next_to(barrier_graph, UP, buff=0.3).shift(LEFT * 3)
        self.play(Create(barrier_graph), Write(barrier_label))

        def wave_incident(x, t):
            return 0.6 * np.exp(-((x + 4 - t) ** 2) / 1.2) * np.cos(6 * (x + 4 - t))

        tracker = ValueTracker(0)
        wave = always_redraw(lambda: axes.plot(
            lambda x: wave_incident(x, tracker.get_value()), x_range=[-6, 6], color=BLUE))

        self.add(wave)
        self.play(tracker.animate.set_value(4), run_time=3, rate_func=linear)

        def transmitted(x, amp):
            return amp * np.exp(-((x - 3) ** 2) / 1.2) * np.cos(6 * (x - 3))

        trans_wave = axes.plot(lambda x: transmitted(x, 0.25), x_range=[1, 6], color=GREEN)
        trans_label = Text("Transmitted (reduced amplitude)", font_size=20, color=GREEN).to_edge(DOWN)
        self.play(Create(trans_wave), Write(trans_label))
        self.wait(2)

````

````python
%%manim -qm VibrationalSpectroscopy

from manim import *
import numpy as np

class VibrationalSpectroscopy(Scene):
    def construct(self):
        title = Text("Vibrational Spectroscopy: Morse Potential Energy Levels", font_size=25).to_edge(UP)
        self.play(Write(title))

        axes = Axes(x_range=[0, 5, 1], y_range=[0, 5, 1], x_length=9, y_length=5.5).shift(DOWN * 0.2)
        self.play(Create(axes))

        def morse(r, D=4, a=1, r0=2):
            return D * (1 - np.exp(-a * (r - r0))) ** 2

        curve = axes.plot(morse, x_range=[0.7, 4.5], color=GREY)
        self.play(Create(curve))

        for n in range(4):
            E = (n + 0.5) - 0.1 * (n + 0.5) ** 2
            hline = DashedLine(axes.c2p(1.2, E), axes.c2p(3.6 - n*0.15, E), color=[BLUE,GREEN,YELLOW,RED][n])
            label = MathTex(f"v={n}", font_size=20, color=[BLUE,GREEN,YELLOW,RED][n]).next_to(hline, RIGHT)
            self.play(Create(hline), Write(label), run_time=0.7)

        note = Text("Levels converge near dissociation energy D -- anharmonic spacing", font_size=18).to_edge(DOWN)
        self.play(Write(note))
        self.wait(2)
````

````python
def checkMagicLineClassName(idx,code):
    matchMagicline=MAGIC_LINE_RE.search(code)
    if not(matchMagicline): 
        print(f"Magic line not present in the code line {idx+1}")
        return None
    else:
        className=CLASS_RE.search(code)
        if not className: 
            print(f"Class Name not present in the code line {idx+1}")
            return None
        else: #check if magic line contains the correct classname
            matchclassName=className.search(matchMagicline)
            if matchclassName:
                return matchMagicline.group(1) and className.group(1)
            else:
                print(f"Class name in the magic line does not match the actual class name at line {idx+1}")
                return None
````
