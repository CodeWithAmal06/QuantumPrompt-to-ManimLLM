add magic line to dataset

````python
with open("input.jsonl", "r",encoding="utf-8") as file:
    lines = file.readlines()
    
for index,line in enumerate(lines):
    if index>=25 and index<=99:
        marker1="class "
        start1=line.find(marker1)
        if start1==-1:
            continue #marker not present in the line
        end1=start1+len(marker1)
        for i in range(end1,len(line)):
            if line[i] == '(':
                classname = line[end1:i]
                break
        insert='%%manim -qm '+classname+'\\n'

        marker="```python"
        start=line.find(marker)
        if start==-1:
            continue
        end=start+len(marker)
        lines[index]=line[:end]+'\\n'+insert+line[end:]
    

with open("input.jsonl",'w', encoding="utf-8") as file:
    file.writelines(lines)
````

````python
import json
import re
import ast

CODE_BLOCK_RE=re.compile(r"```python\n(.*?)\n```",re.DOTALL)
MAGIC_LINE_RE=re.compile(r"%%manim\s+(-\S+\s+)*(\S+\s+)")
CLASS_RE=re.compile(r"class\s+(\S+)\s*\(")
````

````python

def extract_code(assistant_content: str) -> str | None:
    match=CODE_BLOCK_RE.search(assistant_content)
    return match.group(1) if match else None
````

````python
def checkMagicLineClassName(idx,code):
    matchMagicline=MAGIC_LINE_RE.search(code)
    if not(matchMagicline): 
        print(f"Magic line not present in the code line {idx+1}")
        return None
    else:
        className=CLASS_RE.search(code)
        if not className: 
            print(f"Class Name not present in the code line {idx+1}")
            return None
        else: #check if magic line contains the correct classname
            matchclassName=className.search(matchMagicline)
            if matchclassName:
                return matchMagicline.group(1) and className.group(1)
            else:
                print(f"Class name in the magic line does not match the actual class name at line {idx+1}")
                return None
````
